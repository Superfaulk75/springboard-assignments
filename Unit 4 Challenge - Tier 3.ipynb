**2.2. Cleaning the data**

You might find you need to transpose your dataframe, check out what its row indexes are, and reset the index. You  also might find you need to assign the values of the first row to your column headings  . (Hint: recall the .columns feature of DataFrames, as well as the iloc[] method).

```python
# Transpose so boroughs are rows, dates are columns
properties_T = properties.T

# Inspect the transposed dataframe
properties_T.head()
```

```python
# Assign the first row as the column headers (these are the dates/months)
properties_T.columns = properties_T.iloc[0]

# Drop the first row (it's now the header)
properties_T = properties_T.drop(properties_T.index[0])

# Reset the index so borough names become a column
properties_T.reset_index(inplace=True)

properties_T.head()
```

**2.3. Cleaning the data (part 2)**

You might we have to **rename** a couple columns. How do you do this? The clue's pretty bold...

```python
# The 'index' column (borough names) and the first date column need renaming
# Check current column names
print(properties_T.columns[:5])
```

```python
# Rename the borough name column
properties_T.rename(columns={'Unnamed: 0': 'London_Borough', pd.NaT: 'ID'}, inplace=True)

properties_T.head()
```

**2.4.Transforming the data**

Remember what Wes McKinney said about tidy data? 

You might need to **melt** your DataFrame here. 

```python
# Melt the dataframe: keep London_Borough and ID as identifier variables
# All date columns become rows in a 'Month' column, values go into 'Average Price'
dfg = pd.melt(properties_T,
              id_vars=['London_Borough', 'ID'],
              var_name='Month',
              value_name='Average Price')

dfg.head(10)
```

Remember to make sure your column data types are all correct. Average prices, for example, should be floating point numbers... 

```python
# Ensure correct data types
dfg['Average Price'] = pd.to_numeric(dfg['Average Price'], errors='coerce')
dfg['Month'] = pd.to_datetime(dfg['Month'], errors='coerce')

print(dfg.dtypes)
dfg.head()
```

**2.5. Cleaning the data (part 3)**

Do we have an equal number of observations in the ID, Average Price, Month, and London Borough columns? Remember that there are only 32 London Boroughs. How many entries do you have in that column? 

Check out the contents of the London Borough column, and if you find null values, get rid of them however you see fit. 

```python
# Check unique values in London_Borough to spot non-borough entries
print(dfg['London_Borough'].unique())
```

```python
# Check for nulls
print(dfg.isnull().sum())
```

```python
# Drop rows where London_Borough is NaN or not an actual borough
# Non-borough entries include things like 'Unnamed', 'Inner London', 'Outer London', 'LONDON' etc.
non_boroughs = ['Inner London', 'Outer London', 'LONDON', 
                'England', 'United Kingdom', 'South East', 
                'East of England', 'East Midlands', 'West Midlands',
                'Yorkshire and The Humber', 'North West', 'North East',
                'South West', 'Wales', 'Scotland', 'Northern Ireland']

# Drop nulls first
dfg.dropna(subset=['London_Borough'], inplace=True)

# Drop non-borough rows
dfg = dfg[~dfg['London_Borough'].isin(non_boroughs)]

# Also drop any rows where Average Price is NaN
dfg.dropna(subset=['Average Price'], inplace=True)

print(f"Unique boroughs: {dfg['London_Borough'].nunique()}")
print(dfg['London_Borough'].unique())
```

```python
# Final shape check
dfg.shape
```

**3. Modeling**

Consider creating a function that will calculate a ratio of house prices, comparing the price of a house in 2018 to the price in 1998.

Consider calling this function create_price_ratio.

You'd want this function to:
1. Take a filter of dfg, specifically where this filter constrains the London_Borough, as an argument. For example, one admissible argument should be: dfg[dfg['London_Borough']=='Camden'].
2. Get the Average Price for that Borough, for the years 1998 and 2018.
4. Calculate the ratio of the Average Price for 1998 divided by the Average Price for 2018.
5. Return that ratio.

Once you've written this function, you ultimately want to use it to iterate through all the unique London_Boroughs and work out the ratio capturing the difference of house prices between 1998 and 2018.

Bear in mind: you don't have to write a function like this if you don't want to. If you can solve the brief otherwise, then great! 

***Hint***: This section should test the skills you acquired in:
- Python Data Science Toolbox - Part One, all modules

```python
def create_price_ratio(borough_df):
    """
    Calculate the ratio of average house price in 2018 vs 1998
    for a given borough subset of dfg.
    
    A higher ratio means greater price growth.
    """
    price_1998 = borough_df[borough_df['Year'] == 1998]['Average Price'].mean()
    price_2018 = borough_df[borough_df['Year'] == 2018]['Average Price'].mean()
    
    if price_1998 == 0 or np.isnan(price_1998):
        return np.nan
    
    ratio = price_2018 / price_1998
    return ratio
```

```python
# Test the function on Camden
print(f"Camden price ratio (2018/1998): {create_price_ratio(dfg[dfg['London_Borough'] == 'Camden']):.2f}x")
```

```python
# Iterate through all boroughs and compute their price ratios
boroughs = dfg['London_Borough'].unique()

ratios = {}
for borough in boroughs:
    ratio = create_price_ratio(dfg[dfg['London_Borough'] == borough])
    ratios[borough] = ratio

# Convert to a Series and sort
ratio_series = pd.Series(ratios).dropna().sort_values(ascending=False)

print(ratio_series)
```

```python
# Visualize the top 10 boroughs with greatest price increase
top10 = ratio_series.head(10)

plt.figure(figsize=(12, 6))
top10.plot(kind='bar', color='steelblue', edgecolor='black')
plt.title('Top 10 London Boroughs: House Price Growth Ratio (2018 vs 1998)', fontsize=13)
plt.xlabel('Borough')
plt.ylabel('Price Ratio (2018 Price / 1998 Price)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
# Full ranking as a DataFrame
ratio_df = ratio_series.reset_index()
ratio_df.columns = ['London_Borough', 'Price_Ratio_2018_vs_1998']
ratio_df['Price_Ratio_2018_vs_1998'] = ratio_df['Price_Ratio_2018_vs_1998'].round(2)
ratio_df
```
